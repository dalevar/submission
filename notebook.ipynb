## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Kategori produk apa yang menghasilkan total pendapatan (revenue) tertinggi dan bagaimana tren penjualannya secara bulanan selama periode tahun 2017 hingga 2018?

  Keterangan:
  *   **Specific:** Fokus pada analisis kategori produk, pendapatan (revenue), dan tren bulanan.
  *   **Measurable:** Diukur berdasarkan angka pendapatan bersih dan volume penjualan bulanan.
  *   **Action-Oriented:** Membantu tim manajemen persediaan (inventory) dan pemasaran menentukan alokasi anggaran promosi serta stok barang pada kategori prioritas.
  *   **Relevant:** Mengidentifikasi produk pembuat keuntungan (profit driver) utama sangat penting bagi keberlanjutan bisnis e-commerce.
  *   **Time-bound:** Dibatasi pada periode analisis tahun 2017 hingga 2018.


- **Pertanyaan 2:** Berapa rata-rata waktu pengiriman pesanan (delivery time) untuk setiap negara bagian (customer state) dan apakah keterlambatan pengiriman secara signifikan memengaruhi skor ulasan (review score) pelanggan?

  Keterangan:
  * **Specific:** Fokus pada korelasi antara durasi/keterlambatan pengiriman dan skor ulasan berdasarkan lokasi geografis pelanggan (customer state).

  * **Measurable:** Diukur dengan menghitung rata-rata selisih hari pengiriman serta nilai rerata review score (skala 1-5).

  * **Action-Oriented:** Memberikan masukan langsung bagi tim logistik untuk mengevaluasi mitra pengiriman di wilayah dengan tingkat keterlambatan dan ulasan buruk paling tinggi.

  * **Relevant:** Kecepatan pengiriman berpengaruh langsung terhadap tingkat kepuasan dan loyalitas pelanggan.

  * **Time-bound:** Menganalisis seluruh riwayat transaksi pengiriman dan ulasan yang tercatat dalam dataset.


- **Pertanyaan 3:** Metode pembayaran apa yang paling sering digunakan oleh pelanggan berdasarkan jumlah transaksi dan total nilai pembayaran pada setiap kuartal?

  Keterangan:
  * **Specific:** Fokus pada preferensi metode pembayaran pelanggan (seperti kartu kredit, boleto, voucher, dll.) berdasarkan volume dan total nilai transaksi.

  * **Measurable:** Diukur menggunakan total frekuensi penggunaan (jumlah transaksi) dan nominal rupiah/mata uang (payment value).

  * **Action-Oriented:** Membantu tim fintech / operasional untuk mengoptimalkan integrasi sistem pembayaran, menegosiasikan biaya transaksi (fee payment gateway), atau memberikan promo diskon metode pembayaran tertentu.

  * **Relevant:** Memahami tren pembayaran dapat meningkatkan checkout completion rate dan mengurangi kerugian akibat opsi pembayaran yang kurang efisien.

  * **Time-bound:** Dievaluasi secara berkala pada setiap kuartal (quarterly).

## Data Wrangling

### Gathering Data

#### Load df ...

```python
# Memuat seluruh dataset e-commerce
customers_df = pd.read_csv('customers_dataset.csv')

orders_df = pd.read_csv('orders_dataset.csv')
order_items_df = pd.read_csv('order_items_dataset.csv')
order_payments_df = pd.read_csv('order_payments_dataset.csv')
order_reviews_df = pd.read_csv('order_reviews_dataset.csv')

products_df = pd.read_csv('products_dataset.csv')
product_category_name_translation_df = pd.read_csv('product_category_name_translation.csv')

print('Dataset berhasil dimuat')
```

**Insight:** (Opsional)
- Seluruh 7 dataset e-commerce berhasil dimuat ke dalam DataFrame tanpa terjadi error.
- Dataset mencakup entitas utama bisnis seperti transaksi (orders), detail item (order_items), pembayaran (order_payments), ulasan (order_reviews), data pelanggan (customers), katalog produk (products), dan pemetaan bahasa (product_category_name_translation).

### Assessing Data

#### Identifying ... problem

```python
# Menilai kualitas data pada setiap DataFrame
print("--- Memeriksa Missing Value ---")
print("Orders Missing Values:\n", orders_df.isna().sum())
print("\nProducts Missing Values:\n", products_df.isna().sum())
print("\nReviews Missing Values:\n", order_reviews_df.isna().sum())

print("\n--- Memeriksa Duplikasi ---")
print("Orders Duplicates:", orders_df.duplicated().sum())
print("Order Items Duplicates:", order_items_df.duplicated().sum())

print("\n--- Memeriksa Tipe Data ---")
print(orders_df.info())
```

**Steps to Take:**
- Handling Missing Values: Mengisi nilai kosong (missing values) pada kolom product_category_name di DataFrame produk dengan label 'others' serta mengabaikan/menghapus baris yang tidak memiliki tanggal pengiriman (order_delivered_customer_date) saat menganalisis performa waktu pengiriman logistik.
- Tipe Data Inconsistent: Mengubah tipe data seluruh kolom tanggal pada orders_df (seperti order_purchase_timestamp, order_approved_at, order_delivered_carrier_date, order_delivered_customer_date, dan order_estimated_delivery_date) dari tipe object (string) menjadi datetime.
- Menerjemahkan Kategori Produk: Menggabungkan (merge) products_df dengan product_category_name_translation_df agar kategori produk berbahasa Portugis dapat diterjemahkan ke bahasa Inggris untuk memudahkan analisis bisnis.

**Insight:** (Opsional)
- Ditemukan missing values pada orders_df (terutama order_delivered_customer_date sebanyak 2.965 baris)
- Seluruh kolom tanggal di orders_df masih tersimpan sebagai tipe data string (object), sehingga perlu dikonversi ke datetime agar bisa digunakan dalam analisis deret waktu (time-series) dan perhitungan durasi pengiriman.
- Tidak ditemukan adanya data duplikat pada orders_df maupun order_items_df, menunjukkan bahwa entitas data transaksi unik secara konsisten.

### Cleaning Data

#### Fixing ... problem

```python
# 1. Mengubah tipe data kolom tanggal menjadi datetime pada orders_df
datetime_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in datetime_cols:
    orders_df[col] = pd.to_datetime(orders_df[col])

# 2. Menggabungkan nama kategori bahasa Inggris dan mengisi nilai NaN tanpa menggunakan inplace=True
products_df = products_df.merge(product_category_name_translation_df, on='product_category_name', how='left')
products_df['product_category_name_english'] = products_df['product_category_name_english'].fillna('others')

# 3. Menghapus baris yang tidak memiliki tanggal pengiriman pelanggan
orders_clean_df = orders_df.dropna(subset=['order_delivered_customer_date']).copy()

# 4. Menambahkan kolom durasi pengiriman (hari) dan status keterlambatan
orders_clean_df['delivery_time_days'] = (orders_clean_df['order_delivered_customer_date'] - orders_clean_df['order_purchase_timestamp']).dt.days
orders_clean_df['is_delayed'] = orders_clean_df['order_delivered_customer_date'] > orders_clean_df['order_estimated_delivery_date']

print("Proses cleaning data selesai. Tipe data kolom tanggal terverifikasi:")
print(orders_clean_df[datetime_cols].dtypes)
```

**Insight:** (Opsional)
- Seluruh 5 kolom tanggal pada orders_df telah sukses dikonversi dari tipe data object ke datetime64[ns] untuk memfasilitasi kalkulasi durasi pengiriman.

- Missing values pada kolom kategori produk ditangani dengan menggabungkan nama kategori bahasa Inggris dan mengisi nilai kosong menggunakan label 'others'.

- Dua kolom fitur baru (delivery_time_days dan is_delayed) berhasil dibuat untuk mendukung analisis performa ketepatan waktu pengiriman dan dampaknya terhadap kepuasan pelanggan.

### Explore ...

```python
# 1. Menggabungkan dataset utama menjadi satu DataFrame eksplorasi
master_df = orders_clean_df.merge(order_items_df, on='order_id', how='inner')
master_df = master_df.merge(products_df, on='product_id', how='left')
master_df = master_df.merge(customers_df, on='customer_id', how='left')
master_df = master_df.merge(order_payments_df, on='order_id', how='left')
master_df = master_df.merge(order_reviews_df, on='order_id', how='left')

# 2. Eksplorasi Pertanyaan 1: Top Kategori Produk berdasarkan Revenue (2017-2018)
revenue_2017_2018 = master_df[
    (master_df['order_purchase_timestamp'].dt.year >= 2017) &
    (master_df['order_purchase_timestamp'].dt.year <= 2018)
].groupby('product_category_name_english')['price'].sum().sort_values(ascending=False).reset_index()

# 3. Eksplorasi Pertanyaan 2: Performa Pengiriman dan Skor Ulasan per Negara Bagian
delivery_performance = master_df.groupby('customer_state').agg(
    avg_delivery_days=('delivery_time_days', 'mean'),
    avg_review_score=('review_score', 'mean'),
    delay_rate=('is_delayed', 'mean'),
    total_orders=('order_id', 'nunique')
).reset_index().sort_values(by='avg_delivery_days', ascending=False)

# 4. Eksplorasi Pertanyaan 3: Tren Distribusi Metode Pembayaran
payment_distribution = master_df.groupby('payment_type').agg(
    total_transactions=('order_id', 'nunique'),
    total_payment_value=('payment_value', 'sum')
).reset_index().sort_values(by='total_payment_value', ascending=False)

# Menampilkan statistik hasil eksplorasi
print("=== Top 5 Kategori Produk Revenue Terbaik ===")
print(revenue_2017_2018.head())

print("\n=== Performa Pengiriman & Ulasan per Negara Bagian (Top 5 Paling Lama) ===")
print(delivery_performance.head())

print("\n=== Ringkasan Penggunaan Metode Pembayaran ===")
print(payment_distribution)

# Verifikasi ketersediaan kolom sebelum di-export
print("Kolom yang ada di master_df:")
print([col for col in ['price', 'payment_value', 'product_category_name_english'] if col in master_df.columns])

# Simpan ke main_data.csv untuk digunakan dalam pembuatan dashboard streamlit
master_df.to_csv('main_data.csv', index=False)
print("File 'main_data.csv' berhasil diperbarui dengan seluruh kolom lengkap!")
```

**Insight:** (Opsional)
- Kategori produk health_beauty mendominasi total pendapatan tertinggi ( 1272112.46), diikuti oleh watches_gifts (1211631.38) dan bed_bath_table (1091982.21) sepanjang periode 2017–2018.
- Negara bagian dengan rata-rata waktu pengiriman paling lama (seperti RR, AP, dan AM yang mencapai > 26 hari) cenderung memiliki tingkat keterlambatan tinggi serta rata-rata review score yang lebih rendah (< 3,9 pada wilayah RR dan AL).
- Opsi pembayaran menggunakan credit_card menjadi metode yang paling mendominasi transaksi dengan total 74.303 transaksi dan total nilai pembayaran mencapai 15,26 juta BRL, jauh melampaui metode boleto, voucher, dan debit_card.

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
# Visualisasi Pertanyaan 1: Top 5 Kategori Produk Berdasarkan Revenue (2017-2018)
top_5_categories = master_df[
    (master_df['order_purchase_timestamp'].dt.year >= 2017) &
    (master_df['order_purchase_timestamp'].dt.year <= 2018)
].groupby('product_category_name_english')['price'].sum().nlargest(5).reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(
    data=top_5_categories,
    x='price',
    y='product_category_name_english',
    hue='product_category_name_english',
    legend=False,
    palette='Blues_r'
)
plt.title('Top 5 Kategori Produk Berdasarkan Total Pendapatan (2017–2018)', fontsize=14, fontweight='bold')
plt.xlabel('Total Pendapatan (BRL)', fontsize=12)
plt.ylabel('Kategori Produk', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Pertanyaan 2:

```python
# Visualisasi Pertanyaan 2: Hubungan Rata-rata Waktu Pengiriman dengan Skor Ulasan
delivery_by_state = master_df.groupby('customer_state').agg({
    'delivery_time_days': 'mean',
    'review_score': 'mean'
}).reset_index()

plt.figure(figsize=(10, 5))
sns.scatterplot(
    data=delivery_by_state,
    x='delivery_time_days',
    y='review_score',
    color='crimson',
    s=100,
    alpha=0.8
)
plt.title('Pengaruh Rata-Rata Waktu Pengiriman Terhadap Skor Ulasan Pelanggan', fontsize=14, fontweight='bold')
plt.xlabel('Rata-Rata Waktu Pengiriman (Hari)', fontsize=12)
plt.ylabel('Rata-Rata Skor Ulasan (1–5)', fontsize=12)
plt.axvline(x=15, color='gray', linestyle='--', label='Batas Pengiriman 15 Hari')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### Pertanyaan 3:

```python
# Visualisasi Pertanyaan 3: Distribusi Metode Pembayaran Berdasarkan Total Nilai Pembayaran
payment_summary = master_df.groupby('payment_type')['payment_value'].sum().reset_index()
payment_summary = payment_summary.sort_values(by='payment_value', ascending=False)

plt.figure(figsize=(9, 5))
sns.barplot(
    data=payment_summary,
    x='payment_type',
    y='payment_value',
    hue='payment_type',
    legend=False,
    palette='Blues_r'
)
plt.title('Distribusi Total Nilai Pembayaran Berdasarkan Metode Pembayaran', fontsize=14, fontweight='bold')
plt.xlabel('Metode Pembayaran', fontsize=12)
plt.ylabel('Total Nilai Pembayaran (BRL)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Menambahkan label nominal angka di atas bar
for index, row in payment_summary.iterrows():
    plt.text(index, row['payment_value'], f"{row['payment_value']/1e6:.2f}M BRL", color='black', ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

## Analisis Lanjutan (Opsional)

```python
# Analisis Lanjutan: RFM Analysis (Recency, Frequency, Monetary)

# 1. Menentukan tanggal referensi (1 hari setelah transaksi terakhir)
recent_date = master_df['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

# 2. Menghitung nilai Recency, Frequency, dan Monetary per pelanggan unik
rfm_df = master_df.groupby('customer_unique_id').agg({
    'order_purchase_timestamp': lambda x: (recent_date - x.max()).days,
    'order_id': 'nunique',
    'price': 'sum'
}).reset_index()

# 3. Mengubah nama kolom
rfm_df.columns = ['customer_unique_id', 'Recency', 'Frequency', 'Monetary']

# Menampilkan 5 pelanggan teratas berdasarkan nilai Monetary
print("=== Top 5 Pelanggan Berdasarkan Monetary Value ===")
print(rfm_df.sort_values(by='Monetary', ascending=False).head())

# Menampilkan statistik deskriptif RFM
print("\n=== Ringkasan Statistik RFM ===")
print(rfm_df[['Recency', 'Frequency', 'Monetary']].describe())
```

**Insight:** (Opsional)
- Sebagian besar pelanggan platform e-commerce merupakan pembeli satu kali (one-time buyers), ditunjukkan oleh rata-rata transaksi (Frequency) sebesar 1,03 dengan nilai median 1,0.
- Rata-rata pengeluaran pelanggan (Monetary) mencatatkan angka 148,63 BRL per pelanggan unik, di mana pelanggan bernilai tertinggi (top spender) melakukan akumulasi belanja hingga mencapai 13.440 BRL.
